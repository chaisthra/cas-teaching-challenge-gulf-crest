# Notebook 8: Final Recommendation

**To:** Gulf Crest Insurance Company, Executive Leadership
**From:** Actuarial Analysis Team
**Re:** Hurricane Marisol Claims Response, Findings and Recommendations

This notebook closes out the module with a written summary in the register an actuarial team would actually
hand to leadership: what was found, what it means for decisions Gulf Crest needs to make now, and what
should be revisited as more data comes in over the following months. It draws on Notebooks 3 through 7 and
adds no new analysis of its own.

## Background

Hurricane Marisol made landfall as a Category 4 storm on July 18, 2026, and generated 2,021 claims against
Gulf Crest's coastal book of business, on top of the routine claims volume the company handles in any given
period. Thirty days after landfall, most of those claims are still open, and only a small fraction have
gone through a full reserve review. Five analyses were run against this claims population and Gulf Crest's
broader policy and financial data, each addressing a specific operational question.

## Finding 1: Claim Severity Can Be Predicted, With a Trade-Off Between Accuracy and Interpretability

**Notebook 3** compared two ways of predicting how much an individual claim will ultimately cost: a Gamma
GLM, the method taught throughout the CAS syllabus and required for actuarial sign-off, and a gradient
boosting model. Gradient boosting scored somewhat better (MAE $37,500 versus $39,615, R-squared 0.334 versus
0.288), but the GLM's coefficients can be read directly: catastrophe status, total insured value, and cause
of loss all move predicted severity in ways that can be explained to a regulator or an underwriter in a
single sentence each. Gradient boosting's predictions cannot be unpacked the same way.

**Recommendation:** Use the GLM for any severity estimate that needs to be filed, defended, or explained
externally. Gradient boosting is appropriate for internal triage and prioritization, where a modest accuracy
edge matters more than being able to explain any single prediction.

## Finding 2: Fraud Detection Requires an Explicit Choice Between Catching Fraud and Avoiding False Alarms

**Notebook 4** showed that a fraud classifier trained without correcting for class imbalance is useless in
practice (0% recall despite 85% accuracy), and that both class weighting and SMOTE substantially improve
on that baseline. Deliberately setting a recall target and tuning the decision threshold to meet it, rather
than defaulting to a 50% cutoff nobody consciously chose, pushed recall to 90% on held-out data. **Notebook
7** then showed this was a reasoned choice rather than an arbitrary one: the precision/recall trade-off
peaks in F1 terms near a 70% recall target, and Gulf Crest's 80% target trades a modest amount of overall
balance for materially better fraud capture, a defensible choice given that a missed fraud case is more
costly than an extra investigation. Isolation Forest, which needs no fraud labels at all, offers a way to
flag suspicious activity among the 43% of claims that have not yet been formally investigated.

**Recommendation:** Deploy the SMOTE-trained model with the recall-targeted threshold to prioritize SIU
investigation of already-reviewed claims, and use the Isolation Forest flag to surface candidates among
claims that have not yet reached formal review. Revisit the 80% recall target periodically as SIU
investigation capacity changes, since Notebook 7 showed that target is a genuine trade-off, not a fixed
requirement.

## Finding 3: Gulf Crest's Reserve for Hurricane Marisol Should Be Set Near $160 Million, Not Near Zero

**Notebook 5** demonstrated that Gulf Crest's routine-claims payment history cannot be used to project
Hurricane Marisol's ultimate cost: naively applying it produces an estimate near $47 thousand, against a
bottom-up adjuster estimate of $168 million, because routine claims pay out far faster than catastrophe
damage and because only 43 of 2,021 Marisol claims have been formally reviewed so far. Bornhuetter-Ferguson,
blending that bottom-up estimate with the small amount actually paid to date, produced an ultimate estimate
of roughly $160 million. **Notebook 7** confirmed this figure is not fragile: a wide range of assumptions
about how much has been paid so far (2% to 15%) moves the estimate by only about 7% in either direction.

This $160 million is Gulf Crest's gross exposure before any reinsurance recovery, and is not directly
comparable to the $44.5 million booked in company-wide incurred losses for the quarter, which reflects the
company's net, retained position across its entire book, not this cohort of claims specifically.

**Recommendation:** Book reserves for Hurricane Marisol claims against the Bornhuetter-Ferguson estimate,
not the immature paid-to-date figure. Revisit the estimate monthly as more claims complete formal review;
Notebook 7's sensitivity analysis suggests this figure should move only modestly as that happens, and a
larger swing than shown there would be worth investigating rather than assuming away.

## Finding 4: A Simple Priority Score Meaningfully Speeds Up Claims Resolution, With a Fairness Cost Worth Managing

**Notebook 6** showed that ranking Gulf Crest's 2,035 open claims by a transparent score (severity, age,
and catastrophe status) rather than working them oldest-first cuts the dollar-weighted average time to
resolution roughly in half and resolves the largest claims nearly six times faster. That gain came with a
real cost: the smallest claims wait substantially longer under pure priority scoring. Reserving a fixed 20%
of daily capacity for the oldest claims regardless of score recovered most of that fairness at almost no
cost to the efficiency gained. **Notebook 7** showed the approach scales predictably with adjuster staffing,
with no sign of a hidden capacity cliff.

**Recommendation:** Adopt the hybrid triage rule (80% of daily capacity by priority score, 20% reserved for
the oldest remaining claims) rather than either pure approach. Treat adjuster staffing as a straightforward
cost-versus-speed lever, since Notebook 7 found the relationship between the two to be smooth and
predictable across the range tested.

## Summary of Recommendations

1. Use the Gamma GLM for any severity figure that must be explained or defended externally; use gradient
   boosting only for internal prioritization.
2. Deploy the recall-targeted, SMOTE-trained fraud classifier for reviewed claims, and Isolation Forest to
   surface candidates among unreviewed ones. Revisit the 80% recall target as SIU capacity changes.
3. Book Hurricane Marisol reserves at the Bornhuetter-Ferguson estimate of roughly $160 million gross, not
   at the immature paid-to-date figure, and update it monthly as more claims complete formal review.
4. Adopt the hybrid triage rule for claims resource allocation, and treat adjuster staffing levels as a
   predictable cost-versus-speed trade-off rather than a source of hidden risk.

## Limitations and What Would Be Revisited With More Time

This module deliberately used methods on the CAS syllabus (Chain Ladder, Bornhuetter-Ferguson, GLMs) rather
than more advanced techniques (Mack's stochastic model, ODP bootstrapping, ensemble reserving methods) that
a working actuarial team would eventually bring to bear on a catastrophe of this size. The reserve estimate
in particular rests on a single scalar assumption about percent paid so far, which Notebook 7 showed to be a
reasonably stable input, but a full reserving process would also carry an explicit range or confidence
interval around the ultimate estimate rather than the single point estimate produced here. Reinsurance
recoveries, which materially affect Gulf Crest's net financial exposure to Marisol, were also out of scope
and would be the natural next analysis once claims mature further.
